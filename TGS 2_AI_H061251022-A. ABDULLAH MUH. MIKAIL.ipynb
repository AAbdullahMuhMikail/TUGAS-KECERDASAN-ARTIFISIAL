# TOPIK 1 — Pemodelan Anomali Magnetik dari Sumber Sederhana

## A. Dasar konsep

Survei magnetik mengukur variasi medan magnet total yang dapat dipengaruhi oleh perbedaan kemagnetan batuan. Bentuk anomali bergantung pada geometri, kedalaman, orientasi, dan kontras suseptibilitas sumber. Untuk latihan ini, digunakan **model kurva anomali berbentuk dipol yang disederhanakan**. Model tersebut berguna untuk memperlihatkan bentuk respons terhadap jarak, tetapi tidak menggantikan pemodelan magnetik tiga dimensi yang mempertimbangkan arah medan utama dan magnetisasi sumber.

## B. Prompt AI (R–T–C–I–K–O–Q–V)

**R — Role:** Bertindak sebagai asisten praktikum metode magnetik untuk mahasiswa Geofisika.  
**T — Task:** Buat dan jelaskan kurva anomali magnetik sintetis terhadap posisi lintasan.  
**C — Context:** Kurva dipakai untuk memahami perubahan sinyal saat lintasan melewati sumber magnetik terlokalisasi.  
**I — Input:** Posisi lintasan dari −500 hingga 500 m, pusat sumber di sekitar 0 m, amplitudo karakteristik 80 nT dan skala lebar 140 m.  
**K — Constraints:** Nyatakan bahwa model bersifat ilustratif; gunakan satuan meter dan nT; jangan menentukan jenis batuan hanya dari bentuk kurva.  
**O — Output:** Tabel sampel, grafik, lokasi nilai maksimum/minimum, dan interpretasi terbatas.  
**Q — Quality criteria:** Label sumbu jelas, satuan konsisten, data dapat direproduksi.  
**V — Verification:** Pastikan posisi dan nilai ekstrem diperoleh langsung dari array hasil perhitungan.

## C. Implementasi Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Data sintetis untuk model respons magnetik sederhana
x_m = np.linspace(-500, 500, 201)
amplitudo_nT = 80.0
lebar_m = 140.0
anomali_nT = amplitudo_nT * (x_m / lebar_m) * np.exp(0.5 * (1 - (x_m / lebar_m)**2))

df_magnetik = pd.DataFrame({"Posisi_lintasan_m": x_m, "Anomali_magnetik_nT": anomali_nT})
idx_max = int(np.argmax(anomali_nT))
idx_min = int(np.argmin(anomali_nT))

print("Lima sampel pertama:")
display(df_magnetik.head())
print(f"Maksimum: {anomali_nT[idx_max]:.2f} nT pada x = {x_m[idx_max]:.1f} m")
print(f"Minimum : {anomali_nT[idx_min]:.2f} nT pada x = {x_m[idx_min]:.1f} m")

plt.figure(figsize=(9, 4.8))
plt.plot(x_m, anomali_nT, label="Anomali sintetis")
plt.axhline(0, linewidth=0.8)
plt.scatter([x_m[idx_max], x_m[idx_min]], [anomali_nT[idx_max], anomali_nT[idx_min]], label="Nilai ekstrem")
plt.xlabel("Posisi sepanjang lintasan (m)")
plt.ylabel("Anomali magnetik (nT)")
plt.title("Respons Magnetik Sintetis terhadap Posisi")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## D. Pembahasan dan verifikasi

Kurva menghasilkan pasangan lobus positif dan negatif karena fungsi yang digunakan bersifat ganjil terhadap pusat lintasan. Nilai ekstrem menunjukkan bagian kurva dengan respons terbesar secara numerik, bukan otomatis lokasi tepat batas atau pusat suatu tubuh batuan. Dalam survei nyata, bentuk anomali juga dipengaruhi oleh arah magnetisasi, medan magnet utama, topografi, serta proses koreksi dan pemisahan medan regional.

**Pemeriksaan mutu:** posisi maksimum dan minimum dihitung dari data yang diplot; sumbu posisi menggunakan meter dan respons menggunakan nanotesla. Model ini hanya contoh matematis, sehingga tidak digunakan untuk menafsirkan litologi atau kedalaman secara unik.

# TOPIK 2 — Perhitungan Anomali Gravitasi Udara Bebas

## A. Dasar konsep

Gravitasi yang terukur di permukaan dipengaruhi oleh lintang, elevasi, topografi, dan distribusi massa bawah permukaan. Koreksi udara bebas memperhitungkan perubahan gravitasi akibat perbedaan elevasi pengamatan terhadap datum, tanpa memasukkan efek massa batuan di antara titik dan datum. Dalam pendekatan sederhana, besar koreksi udara bebas sekitar **0,3086 mGal per meter elevasi**. Tanda koreksi yang dipakai harus mengikuti konvensi pemrosesan yang digunakan; di sini nilai koreksi dihitung sebagai besaran positif terhadap elevasi.

## B. Prompt AI (R–T–C–I–K–O–Q–V)

**R:** Asisten praktikum metode gravitasi.  
**T:** Hitung koreksi udara bebas dan tunjukkan pengaruh elevasi terhadap nilai koreksi.  
**C:** Data mewakili titik-titik pengamatan hipotetis dengan elevasi berbeda.  
**I:** Elevasi 25, 60, 95, 130, dan 165 m; gradien koreksi 0,3086 mGal/m.  
**K:** Pertahankan input, gunakan satuan m dan mGal, serta bedakan nilai koreksi dari anomali Bouguer lengkap.  
**O:** Tabel hasil, grafik, dan penjelasan pola.  
**Q:** Perhitungan transparan dan satuan ditampilkan.  
**V:** Verifikasi bahwa koreksi meningkat linear seiring elevasi pada konvensi yang dipakai.

## C. Implementasi Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

elevasi_m = np.array([25, 60, 95, 130, 165], dtype=float)
gradien_mGal_per_m = 0.3086
koreksi_udara_bebas_mGal = gradien_mGal_per_m * elevasi_m

df_grav = pd.DataFrame({
    "Elevasi_m": elevasi_m,
    "Gradien_mGal_per_m": gradien_mGal_per_m,
    "Koreksi_udara_bebas_mGal": koreksi_udara_bebas_mGal
})
display(df_grav.round(4))

plt.figure(figsize=(8.5, 4.6))
plt.plot(elevasi_m, koreksi_udara_bebas_mGal, "o-", label="Koreksi udara bebas")
plt.xlabel("Elevasi titik pengamatan (m)")
plt.ylabel("Koreksi udara bebas (mGal)")
plt.title("Hubungan Elevasi dan Koreksi Udara Bebas")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("Perubahan koreksi dari titik terendah ke tertinggi:",
      f"{koreksi_udara_bebas_mGal[-1] - koreksi_udara_bebas_mGal[0]:.3f} mGal")
```

## D. Pembahasan dan verifikasi

Grafik berupa garis lurus karena koreksi dihitung menggunakan faktor konstan dikalikan elevasi. Pada konvensi latihan ini, titik yang lebih tinggi memperoleh besaran koreksi udara bebas lebih besar. Koreksi ini **bukan** anomali gravitasi Bouguer lengkap: analisis Bouguer memerlukan antara lain gravitasi observasi, gravitasi normal, elevasi, densitas reduksi, dan—bergantung kebutuhan—koreksi medan.

**Pemeriksaan mutu:** hasil dihitung dari persamaan sederhana \(C_{FA}=0{,}3086h\), dengan \(h\) dalam meter dan \(C_{FA}\) dalam mGal. Saat menerapkan data lapangan, tanda koreksi dan datum harus mengikuti standar pengolahan yang dipilih.

# TOPIK 3 — Analisis Frekuensi Sinyal Geofisika dengan FFT

## A. Dasar konsep

Sinyal geofisika dapat terdiri atas beberapa komponen frekuensi. Transformasi Fourier cepat (*Fast Fourier Transform*, FFT) membantu melihat kandungan frekuensi dominan pada sinyal. Dalam latihan ini dibuat sinyal sintetis yang mengandung dua sinusoid dan sedikit derau. Spektrum yang diperoleh memperlihatkan komponen frekuensi, tetapi interpretasi sinyal lapangan perlu memperhatikan frekuensi sampling, aliasing, jendela waktu, dan derau instrumen.

## B. Prompt AI (R–T–C–I–K–O–Q–V)

**R:** Asisten pengolahan sinyal geofisika.  
**T:** Bangun sinyal sintetis, hitung spektrum amplitudo satu sisi dengan FFT, dan identifikasi puncak frekuensi.  
**C:** Contoh ini menggambarkan pemisahan komponen periodik dari sinyal pengamatan.  
**I:** Frekuensi sampling 100 Hz, durasi 10 s, komponen 4 Hz dan 12 Hz.  
**K:** Gunakan frekuensi Nyquist untuk memeriksa batas spektrum; nyatakan sinyal sebagai sintetis.  
**O:** Grafik domain waktu, spektrum frekuensi, dan frekuensi puncak.  
**Q:** Skala frekuensi benar dan sumbu diberi satuan.  
**V:** Puncak spektrum harus berada di sekitar frekuensi yang ditanam dalam sinyal.

## C. Implementasi Python

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 100.0  # frekuensi sampling (Hz)
durasi_s = 10.0
t = np.arange(0, durasi_s, 1/fs)
rng = np.random.default_rng(27)
sinyal = (1.2*np.sin(2*np.pi*4*t) +
          0.6*np.sin(2*np.pi*12*t) +
          0.25*rng.normal(size=t.size))

N = len(sinyal)
spektrum = np.fft.rfft(sinyal)
frekuensi = np.fft.rfftfreq(N, d=1/fs)
amplitudo = 2*np.abs(spektrum)/N
amplitudo[0] /= 2
if N % 2 == 0:
    amplitudo[-1] /= 2

# Cari beberapa puncak terbesar selain komponen DC
mask = frekuensi > 0
indeks = np.where(mask)[0]
top_idx = indeks[np.argsort(amplitudo[mask])[-5:][::-1]]
print("Lima puncak amplitudo terbesar:")
for i in top_idx:
    print(f"{frekuensi[i]:5.2f} Hz  | amplitudo relatif = {amplitudo[i]:.3f}")

fig, ax = plt.subplots(2, 1, figsize=(9, 7))
ax[0].plot(t[:400], sinyal[:400])
ax[0].set_xlabel("Waktu (s)")
ax[0].set_ylabel("Amplitudo relatif")
ax[0].set_title("Potongan Sinyal Sintetis dalam Domain Waktu")
ax[0].grid(True, alpha=0.3)

ax[1].plot(frekuensi, amplitudo)
ax[1].set_xlim(0, 25)
ax[1].set_xlabel("Frekuensi (Hz)")
ax[1].set_ylabel("Amplitudo relatif")
ax[1].set_title("Spektrum Amplitudo Satu Sisi")
ax[1].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Frekuensi Nyquist = {fs/2:.1f} Hz")
```

## D. Pembahasan dan verifikasi

Spektrum seharusnya menunjukkan puncak dekat 4 Hz dan 12 Hz, karena dua komponen tersebut dimasukkan ke dalam sinyal awal. Derau menambahkan energi pada banyak frekuensi sehingga spektrum tidak hanya terdiri dari dua garis ideal. Frekuensi Nyquist pada sampling 100 Hz adalah 50 Hz; komponen di atas batas tersebut tidak dapat direpresentasikan secara benar tanpa penanganan khusus.

**Pemeriksaan mutu:** jarak antarbin frekuensi ditentukan oleh durasi rekaman, sedangkan batas frekuensi oleh laju sampling. Karena sinyal sintetis diketahui, puncak spektrum dapat dibandingkan langsung dengan frekuensi masukan.

# TOPIK 4 — Estimasi Hiposenter 2D dengan Grid Search Sederhana

## A. Dasar konsep

Penentuan hiposenter menggunakan waktu tiba gelombang di beberapa stasiun untuk mencari lokasi sumber yang paling sesuai dengan observasi. Dalam kasus nyata, model kecepatan, kedalaman, fase P dan S, ketidakpastian picking, serta geometri stasiun sangat memengaruhi hasil. Notebook ini menggunakan model **dua dimensi dan kecepatan konstan** untuk mendemonstrasikan prinsip minimisasi residual waktu tiba. Data sintetis dibuat dari sumber yang diketahui, kemudian ditambahkan sedikit gangguan kecil.

## B. Prompt AI (R–T–C–I–K–O–Q–V)

**R:** Asisten praktikum seismologi.  
**T:** Estimasi posisi sumber 2D menggunakan pencarian grid berdasarkan residual waktu tiba.  
**C:** Demonstrasi konsep lokasi sumber dengan jaringan stasiun sederhana.  
**I:** Koordinat stasiun, waktu tiba sintetis, kecepatan rambat konstan 5 km/s.  
**K:** Gunakan satuan km dan s; jelaskan asumsi kecepatan konstan dan bahwa solusi 2D tidak mengestimasi kedalaman.  
**O:** Tabel stasiun, lokasi estimasi, residual, dan peta geometri.  
**Q:** Lokasi hasil berada dalam domain pencarian dan residual dihitung konsisten.  
**V:** Bandingkan hasil estimasi dengan lokasi sumber yang digunakan untuk membuat data sintetis.

## C. Implementasi Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Jaringan stasiun (km) dan sumber sintetis
stasiun = np.array([[-10, -5], [0, 10], [12, 4], [8, -12], [-8, 9]], dtype=float)
sumber_asli = np.array([2.5, -1.5])  # hanya diketahui karena data dibuat secara sintetis
v_km_s = 5.0

jarak_asli = np.sqrt(((stasiun - sumber_asli)**2).sum(axis=1))
rng = np.random.default_rng(10)
waktu_obs_s = jarak_asli / v_km_s + rng.normal(0, 0.015, size=len(stasiun))

df_stasiun = pd.DataFrame({
    "Stasiun": [f"ST{i+1}" for i in range(len(stasiun))],
    "X_km": stasiun[:, 0],
    "Y_km": stasiun[:, 1],
    "Waktu_tiba_s": waktu_obs_s
})
display(df_stasiun.round(4))

# Grid search: pilih titik yang meminimalkan jumlah kuadrat residual waktu
gx = np.linspace(-15, 15, 121)
gy = np.linspace(-15, 15, 121)
XX, YY = np.meshgrid(gx, gy)
misfit = np.zeros_like(XX)

for j in range(len(gy)):
    for i in range(len(gx)):
        kandidat = np.array([XX[j, i], YY[j, i]])
        t_pred = np.sqrt(((stasiun - kandidat)**2).sum(axis=1)) / v_km_s
        misfit[j, i] = np.sum((waktu_obs_s - t_pred)**2)

jmin, imin = np.unravel_index(np.argmin(misfit), misfit.shape)
sumber_estimasi = np.array([XX[jmin, imin], YY[jmin, imin]])
t_estimasi = np.sqrt(((stasiun - sumber_estimasi)**2).sum(axis=1)) / v_km_s
residual = waktu_obs_s - t_estimasi

print(f"Lokasi sumber sintetis : X={sumber_asli[0]:.2f} km, Y={sumber_asli[1]:.2f} km")
print(f"Lokasi hasil grid search: X={sumber_estimasi[0]:.2f} km, Y={sumber_estimasi[1]:.2f} km")
print(f"RMSE waktu tiba: {np.sqrt(np.mean(residual**2)):.4f} s")
display(pd.DataFrame({"Stasiun": df_stasiun["Stasiun"], "Waktu_obs_s": waktu_obs_s,
                      "Waktu_pred_s": t_estimasi, "Residual_s": residual}).round(4))

plt.figure(figsize=(7.5, 6))
plt.scatter(stasiun[:, 0], stasiun[:, 1], marker="^", s=80, label="Stasiun")
plt.scatter(*sumber_asli, marker="*", s=180, label="Sumber sintetis")
plt.scatter(*sumber_estimasi, marker="x", s=100, label="Estimasi grid")
for i, (sx, sy) in enumerate(stasiun):
    plt.text(sx+0.3, sy+0.3, f"ST{i+1}")
plt.xlabel("X (km)")
plt.ylabel("Y (km)")
plt.title("Geometri Stasiun dan Estimasi Hiposenter 2D")
plt.axis("equal")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## D. Pembahasan dan verifikasi

Grid search menguji banyak kandidat koordinat dan memilih lokasi dengan jumlah kuadrat residual waktu terkecil. Nilai RMSE merangkum perbedaan antara waktu tiba observasi sintetis dan waktu yang diprediksi dari lokasi estimasi. Karena waktu observasi diberi gangguan acak, hasil tidak harus tepat sama dengan lokasi sumber asal, tetapi seharusnya berada di sekitarnya.

**Keterbatasan:** model ini mengasumsikan kecepatan seragam, memakai waktu tempuh langsung, dan mencari lokasi horizontal saja. Lokasi hiposenter lapangan membutuhkan model kecepatan yang layak, penentuan fase yang teliti, serta evaluasi ketidakpastian; hasil ilustratif ini tidak boleh dianggap sebagai solusi gempa nyata.

# TOPIK 5 — Deteksi Anomali pada Data Geofisika dengan Skor-Z

## A. Dasar konsep

Data survei geofisika dapat mengandung nilai yang menyimpang karena respons geologi lokal, gangguan lingkungan, atau kesalahan pengukuran. Salah satu pemeriksaan awal yang sederhana adalah skor-z, yaitu selisih suatu nilai dari rata-rata yang dinyatakan dalam satuan simpangan baku. Skor-z dapat menandai nilai yang perlu diperiksa, tetapi tidak membuktikan bahwa nilai tersebut salah atau berasal dari anomali geologi.

## B. Prompt AI (R–T–C–I–K–O–Q–V)

**R:** Asisten kendali mutu data geofisika.  
**T:** Hitung skor-z data sintetis, tandai nilai dengan nilai absolut skor-z di atas 2, dan visualisasikan hasilnya.  
**C:** Analisis merupakan tahap skrining awal sebelum interpretasi geologi.  
**I:** Deret 20 pengukuran sintetis dalam satuan mGal.  
**K:** Jangan menghapus titik otomatis; tampilkan titik yang ditandai dan tekankan perlunya pemeriksaan lapangan.  
**O:** Statistik ringkas, tabel skor-z, grafik, dan catatan keterbatasan.  
**Q:** Ambang dijelaskan dan hasil dapat direproduksi.  
**V:** Periksa apakah nilai yang ditandai benar-benar memiliki skor-z absolut di atas ambang.

## C. Implementasi Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

nilai_mGal = np.array([
    12.1, 12.4, 11.9, 12.3, 12.0, 12.2, 12.5, 12.1, 11.8, 12.4,
    12.2, 12.0, 12.3, 12.1, 12.4, 12.2, 12.0, 12.3, 16.8, 11.9
], dtype=float)

rata = nilai_mGal.mean()
simpangan_baku = nilai_mGal.std(ddof=1)
skor_z = (nilai_mGal - rata) / simpangan_baku
ambang = 2.0
ditandai = np.abs(skor_z) > ambang

df_qc = pd.DataFrame({
    "Titik": np.arange(1, len(nilai_mGal)+1),
    "Nilai_mGal": nilai_mGal,
    "Skor_z": skor_z,
    "Perlu_diperiksa": ditandai
})
display(df_qc.round(3))
print(f"Rata-rata = {rata:.3f} mGal")
print(f"Simpangan baku sampel = {simpangan_baku:.3f} mGal")
print(f"Jumlah titik yang ditandai (|z| > {ambang:g}): {ditandai.sum()}")

plt.figure(figsize=(9, 4.8))
plt.plot(df_qc["Titik"], nilai_mGal, "o-", label="Pengukuran")
plt.scatter(df_qc.loc[ditandai, "Titik"], df_qc.loc[ditandai, "Nilai_mGal"],
            s=110, marker="x", label="Ditandai untuk pemeriksaan")
plt.axhline(rata, linestyle="--", label="Rata-rata")
plt.xlabel("Nomor titik pengukuran")
plt.ylabel("Nilai pengamatan (mGal)")
plt.title("Skrining Awal Nilai Menyimpang dengan Skor-Z")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

assert np.all(np.abs(skor_z[ditandai]) > ambang)
print("Verifikasi ambang skor-z berhasil.")
```

## D. Pembahasan dan verifikasi

Titik dengan skor-z absolut lebih besar dari 2 ditandai sebagai kandidat pemeriksaan. Pada data sintetis ini terdapat satu nilai yang sengaja dibuat jauh lebih besar daripada sebagian besar pengukuran. Namun, prosedur skor-z tidak dapat menentukan penyebab nilai tersebut. Nilai yang berbeda dapat merupakan kesalahan pencatatan, gangguan alat, atau respons geologi yang sah.

**Pemeriksaan mutu:** kode menampilkan rata-rata, simpangan baku sampel, skor-z setiap titik, dan status penandaan. Pernyataan `assert` memeriksa bahwa seluruh titik yang ditandai memang melewati ambang. Jangan menghapus data hanya berdasarkan satu aturan statistik tanpa meninjau metadata, catatan lapangan, dan metode pengukuran.
