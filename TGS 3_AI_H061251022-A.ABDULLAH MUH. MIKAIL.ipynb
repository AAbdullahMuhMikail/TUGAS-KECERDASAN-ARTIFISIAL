## 1. Pemeriksaan Kualitas Data Stasiun dalam Format CSV

### Tujuan
Menilai apakah tabel stasiun memiliki struktur dan informasi waktu yang cukup konsisten untuk digunakan pada tahap pengolahan berikutnya. Pemeriksaan ini bersifat diagnostik dan tidak mengubah data sumber.

### Dasar Singkat
Data tabular perlu diperiksa sebelum dianalisis. Kolom yang tidak konsisten, pencatatan waktu yang berulang, atau nilai kosong dapat menyebabkan hasil rekap menjadi keliru. Karena itu, laporan pemeriksaan sebaiknya menunjukkan masalah yang ditemukan tanpa memperbaikinya secara otomatis.

### Masukan dan Ketentuan
- `CSV_PATH`: lokasi berkas CSV.
- `TIME_COLUMN`: nama kolom yang berisi tanggal atau waktu.
- `REQUIRED_COLUMNS`: daftar kolom yang wajib ada.
- `UNITS`: informasi satuan untuk variabel yang relevan.
- Berkas sumber diperlakukan hanya-baca.
- Nilai yang tidak valid dilaporkan, bukan langsung dihapus atau diganti.
- Jangan menarik kesimpulan ilmiah yang tidak dapat dibuktikan dari pemeriksaan tabel.

### Instruksi dan Alur Kerja
**Prompt yang dapat digunakan:**

> Tinjau berkas `[CSV_PATH]` untuk mengetahui kesiapan datanya. Cocokkan kolom yang tersedia dengan daftar kolom wajib, periksa tipe data, rentang waktu, nilai kosong, dan baris dengan timestamp yang sama. Berikan ringkasan temuan serta catat setiap ketidaksesuaian. Jangan mengubah berkas asli dan jangan menganggap data telah bersih hanya karena berkas berhasil dibuka.

**Urutan pemeriksaan:**
1. Baca tabel tanpa menulis ulang berkas sumber.
2. Cocokkan nama kolom aktual dengan skema yang dibutuhkan.
3. Periksa apakah kolom waktu dapat ditafsirkan sebagai tanggal/waktu.
4. Catat waktu paling awal dan paling akhir yang terbaca.
5. Hitung nilai kosong per kolom dan kemunculan timestamp berulang.
6. Susun tabel status pemeriksaan beserta catatan untuk setiap temuan.

### Rencana Implementasi
Python dengan `pandas` dapat digunakan untuk membaca tabel dan membuat ringkasan. Validasi waktu dapat dilakukan setelah kolom waktu dikenali. Nama kolom yang dipakai perlu disesuaikan dengan struktur CSV sebenarnya; jangan menganggap skema tertentu pasti tersedia.

### Hasil yang Diharapkan
- Laporan pemeriksaan dalam format Markdown.
- Ringkasan jumlah baris dan kolom.
- Daftar kolom yang ditemukan atau tidak tersedia.
- Rentang waktu, jumlah nilai kosong, dan jumlah timestamp berulang.
- Catatan status proses, termasuk pesan kesalahan bila pembacaan gagal.

### Validasi dan Bukti
Pastikan jumlah baris yang dilaporkan cocok dengan berkas yang dibaca dan berkas sumber tidak berubah. Jika program belum dijalankan, status harus ditulis **Not yet executed**. Status **Executed** hanya digunakan jika ada bukti bahwa proses benar-benar telah berjalan.

## 2. Penataan dan Inventarisasi Berkas GeoJSON

### Tujuan
Merancang proses pemeriksaan dan penataan data GeoJSON agar informasi fitur lebih mudah ditinjau, dengan tetap mempertahankan geometri dan atribut yang memiliki dasar jelas.

### Dasar Singkat
GeoJSON menyimpan objek geografis dalam struktur JSON. Pemeriksaan awal mencakup bentuk struktur, daftar fitur, jenis geometri, koordinat, serta atribut. Data yang berhasil dibaca belum tentu memiliki geometri yang benar secara topologis atau informasi sistem koordinat yang lengkap.

### Masukan dan Ketentuan
- `GEOJSON_PATH`: lokasi GeoJSON awal.
- Lokasi tujuan untuk hasil GeoJSON, tabel inventaris, dan ringkasan pemeriksaan.
- Jangan menghapus fitur atau atribut hanya karena tampak tidak biasa.
- Jangan menebak sistem koordinat yang tidak dinyatakan oleh sumber.
- Jangan menukar urutan koordinat tanpa alasan yang dapat diverifikasi.
- Catat fitur yang tidak dapat diperiksa atau diproses.

### Instruksi dan Alur Kerja
**Prompt yang dapat digunakan:**

> Periksa `[GEOJSON_PATH]` dan rancang keluaran berupa GeoJSON tertata, tabel inventaris fitur, serta ringkasan QA/QC. Tinjau struktur JSON, jumlah fitur, tipe geometri, kelengkapan atribut, koordinat yang tidak masuk akal, geometri kosong, dan informasi CRS yang tersedia. Pisahkan temuan yang dapat dipastikan dari hal-hal yang masih memerlukan konfirmasi. Jangan menyatakan file keluaran sudah terbentuk sebelum proses dijalankan.

**Tahapan kerja:**
1. Buka berkas dan pastikan sintaks JSON dapat dibaca.
2. Periksa apakah struktur utama dan koleksi fitur sesuai harapan.
3. Rekap jenis geometri serta atribut yang digunakan tiap fitur.
4. Tandai geometri kosong, koordinat bermasalah, dan atribut yang hilang.
5. Tinjau informasi CRS dan urutan koordinat berdasarkan sumber data.
6. Terapkan penataan hanya pada aspek yang aturannya jelas.
7. Ekspor inventaris dan ringkasan, lalu catat fitur yang gagal diproses.

### Rencana Implementasi
Modul `json` dapat digunakan untuk meninjau struktur dasar, sedangkan pustaka geospasial yang tersedia dapat membantu pemeriksaan geometri dan ekspor. Pustaka serta versinya perlu dicatat setelah lingkungan kerja diperiksa.

### Hasil yang Diharapkan
- GeoJSON hasil penataan, jika proses dijalankan.
- CSV inventaris fitur berisi identitas, tipe geometri, dan ringkasan atribut.
- JSON atau Markdown yang merangkum temuan pemeriksaan.
- Catatan peringatan dan fitur yang memerlukan tinjauan manual.

### Validasi dan Bukti
Bandingkan jumlah fitur sebelum dan sesudah penataan. Pastikan atribut penting tetap tersedia dan struktur hasil dapat dibaca kembali. Perubahan jumlah fitur harus dijelaskan; jangan menganggap kesamaan jumlah fitur sebagai satu-satunya bukti bahwa semua geometri valid.

## 3. Simulasi Perambatan Panas pada Batang Satu Dimensi

### Tujuan
Menyusun recipe untuk memperkirakan perubahan suhu di sepanjang satu dimensi selama interval waktu tertentu berdasarkan kondisi awal, sifat termal bahan, kondisi batas, dan resolusi perhitungan.

### Dasar Singkat
Model difusi panas satu dimensi dapat dituliskan sebagai:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

dengan `T` sebagai suhu, `t` sebagai waktu, `x` sebagai posisi, dan `α` sebagai difusivitas termal. Pada skema beda hingga eksplisit sederhana, parameter kestabilan biasanya dinyatakan:

\[
r=\frac{\alpha\Delta t}{(\Delta x)^2}
\]

Untuk skema eksplisit standar satu dimensi, syarat umum kestabilan adalah `r ≤ 1/2`. Ketentuan tersebut harus ditinjau kembali jika metode numerik yang digunakan berbeda.

### Masukan dan Ketentuan
- `INITIAL_TEMPERATURE`: profil suhu pada awal simulasi.
- `ALPHA`: difusivitas termal beserta satuannya.
- `BOUNDARY_CONDITIONS`: kondisi pada kedua ujung domain.
- `DX`: jarak antartitik grid.
- `DT`: interval waktu komputasi.
- `TOTAL_TIME`: lama simulasi.
- Seluruh satuan harus konsisten.
- Jangan mengklaim hasil telah konvergen tanpa uji yang mendukung.

### Instruksi dan Alur Kerja
**Prompt yang dapat digunakan:**

> Rancang simulasi difusi panas satu dimensi menggunakan profil awal `[INITIAL_TEMPERATURE]`, difusivitas `[ALPHA]`, kondisi batas `[BOUNDARY_CONDITIONS]`, jarak grid `[DX]`, interval waktu `[DT]`, dan durasi `[TOTAL_TIME]`. Periksa konsistensi satuan serta kestabilan numerik sebelum iterasi. Hasil yang direncanakan mencakup profil suhu pada beberapa waktu, grafik perubahan suhu, tabel parameter, dan catatan pelaksanaan. Nyatakan keterbatasan model dan jangan mengklaim program telah dijalankan tanpa bukti.

**Tahapan kerja:**
1. Verifikasi parameter dan satuan masukan.
2. Bangun grid posisi dan grid waktu.
3. Terapkan profil awal serta kondisi batas.
4. Hitung parameter kestabilan dan hentikan proses jika syarat metode tidak terpenuhi.
5. Lakukan iterasi numerik apabila implementasi tersedia.
6. Simpan profil suhu pada waktu yang dipilih.
7. Bandingkan hasil dengan kasus sederhana atau solusi analitik bila tersedia.
8. Dokumentasikan parameter, peringatan, dan status eksekusi.

### Rencana Implementasi
`numpy` dapat dipakai untuk operasi numerik dan penyimpanan grid, sedangkan `matplotlib` dapat digunakan untuk menggambar profil suhu. Implementasi perlu menyatakan skema beda hingga yang dipilih dan cara penerapan kondisi batas.

### Hasil yang Diharapkan
- Tabel suhu terhadap posisi untuk waktu-waktu terpilih.
- Grafik profil suhu dan perubahan suhu selama simulasi.
- Ringkasan parameter serta pemeriksaan kestabilan.
- Catatan eksekusi yang memuat status dan peringatan.

### Validasi dan Bukti
Periksa kestabilan, kondisi batas, konsistensi dimensi, serta perilaku hasil pada kasus uji sederhana. Perubahan resolusi ruang dan waktu dapat digunakan untuk menilai sensitivitas hasil, tetapi tidak boleh disebut sebagai bukti konvergensi tanpa analisis yang sesuai.

## 4. Pemetaan dan Pemeriksaan Kecepatan Angin dari Data NetCDF

### Tujuan
Membuat alur pengolahan untuk membaca komponen angin zonal dan meridional dari data NetCDF, menghitung kecepatan angin, lalu menampilkan sebarannya pada peta. Recipe ini berbeda dari tiga recipe sebelumnya karena berfokus pada data meteorologi berbasis grid.

### Dasar Singkat
Komponen angin `u` menunjukkan arah timur–barat, sedangkan `v` menunjukkan arah utara–selatan. Besar kecepatan angin horizontal dapat dihitung dengan:

\[
V=\sqrt{u^2+v^2}
\]

Perhitungan harus memperhatikan satuan data, dimensi array, koordinat bujur-lintang, dan indeks waktu yang dipilih. Nilai rata-rata beberapa waktu tidak sama dengan kecepatan pada satu waktu tertentu; metode agregasi perlu dijelaskan.

### Masukan dan Ketentuan
- `NETCDF_PATH`: lokasi berkas NetCDF.
- Nama variabel bujur, lintang, komponen `u`, dan komponen `v`.
- Indeks waktu atau rentang waktu yang akan dianalisis.
- Informasi satuan dan metadata variabel jika tersedia.
- Jangan mengasumsikan nama variabel atau urutan dimensi tanpa memeriksa isi berkas.
- Data sumber tidak boleh ditimpa.
- Area tanpa data harus dipertahankan sebagai nilai kosong, bukan diubah menjadi nol tanpa alasan.

### Instruksi dan Alur Kerja
**Prompt yang dapat digunakan:**

> Buat workflow untuk memeriksa dan memetakan kecepatan angin dari `[NETCDF_PATH]`. Identifikasi variabel koordinat, komponen angin, dimensi waktu, serta satuan berdasarkan metadata. Setelah memastikan bentuk array dan indeks waktu, hitung besar kecepatan dari komponen `u` dan `v`. Tampilkan peta dengan koordinat geografis dan legenda yang menyebutkan satuan. Simpan ringkasan pemeriksaan dan hasil hanya setelah proses berhasil dilakukan. Jangan mengarang nilai yang tidak terdapat dalam data.

**Tahapan kerja:**
1. Baca metadata NetCDF dan daftar variabel yang tersedia.
2. Periksa ukuran setiap dimensi dan urutan dimensi variabel.
3. Ambil komponen `u` dan `v` pada waktu atau periode yang dipilih.
4. Pastikan kedua komponen memiliki bentuk array dan satuan yang sesuai.
5. Hitung kecepatan angin menggunakan persamaan yang ditetapkan.
6. Periksa nilai kosong dan rentang hasil untuk menemukan kemungkinan anomali.
7. Buat peta berdasarkan koordinat bujur dan lintang.
8. Tambahkan judul, label satuan, skala warna, dan informasi periode data.
9. Simpan gambar dan ringkasan proses serta catat kegagalan jika ada.

### Rencana Implementasi
Python dapat menggunakan `xarray` atau `netCDF4` untuk membaca NetCDF, `numpy` untuk menghitung kecepatan, dan `matplotlib` atau `PyGMT` untuk visualisasi. Pilihan pustaka harus mengikuti lingkungan yang benar-benar tersedia. Apabila menggunakan rata-rata waktu, jelaskan apakah yang dirata-ratakan adalah komponen angin terlebih dahulu atau besar kecepatan pada setiap waktu karena kedua cara tersebut dapat memberikan hasil berbeda.

### Hasil yang Diharapkan
1. Peta kecepatan angin dengan legenda dan satuan.
2. Ringkasan nama variabel, dimensi, rentang waktu, dan satuan yang digunakan.
3. File tabel atau array hasil kecepatan jika diperlukan.
4. Catatan QA/QC mengenai nilai kosong dan rentang hasil.
5. Run record yang dibuat hanya jika proses benar-benar dijalankan.

### Pemeriksaan dan Kriteria Keberhasilan
- Koordinat dan dimensi cocok dengan data variabel.
- Komponen `u` dan `v` menggunakan satuan yang sepadan.
- Kecepatan tidak bernilai negatif, kecuali nilai kosong yang direpresentasikan secara khusus.
- Peta menampilkan area, waktu, dan satuan dengan jelas.
- Hasil dapat ditelusuri kembali ke variabel dan indeks waktu sumber.

### Keterbatasan
Peta hanya mewakili periode dan resolusi spasial yang tersedia dalam berkas. Interpretasi pola angin perlu mempertimbangkan sumber data, metode agregasi, serta kualitas metadata. Peta kecepatan saja tidak menunjukkan arah angin; arah perlu dihitung dan divisualisasikan secara terpisah jika dibutuhkan.

### Ringkasan Recipe
Recipe ini menghubungkan pemeriksaan metadata, pengolahan komponen angin, perhitungan besar kecepatan, dan visualisasi geografis. Fokus utamanya adalah menjaga agar hasil peta sesuai dengan variabel serta periode yang benar-benar dipilih, dengan satuan dan keterbatasan yang terdokumentasi.
