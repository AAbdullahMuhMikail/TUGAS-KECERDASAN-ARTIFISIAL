# PROMPT 1 — PENENTUAN HIPOSENTER GEMPA

## 1. Narasi Permasalahan

Penentuan hiposenter merupakan salah satu bagian penting dalam analisis seismologi karena posisi sumber gempa tidak hanya dinyatakan berdasarkan koordinat episenter, tetapi juga kedalaman sumbernya. Penentuan hiposenter dapat dilakukan dengan memanfaatkan waktu tiba gelombang P dan S pada beberapa stasiun seismik. Perbedaan waktu tiba tersebut memberikan informasi mengenai jarak antara stasiun dan sumber gempa.

Dalam pengolahan data gempa, jumlah stasiun, kualitas picking gelombang P dan S, serta model kecepatan yang digunakan dapat memengaruhi hasil lokasi hiposenter. Oleh karena itu, AI dapat digunakan sebagai alat bantu untuk mengorganisasi data, melakukan perhitungan jarak dan residual waktu, serta membantu mengevaluasi hasil penentuan lokasi sumber gempa.

## 2. Struktur Good Prompt

### R — Role

Anda berperan sebagai seorang **seismolog yang membimbing mahasiswa S1 Geofisika dalam menentukan lokasi hiposenter gempa berdasarkan data waktu tiba gelombang seismik**.

### T — Task

Tentukan perkiraan koordinat hiposenter suatu gempa berdasarkan data koordinat stasiun, waktu tiba gelombang P dan S, serta model kecepatan gelombang yang diberikan.

### C — Context

Data berasal dari beberapa stasiun seismik yang merekam kejadian gempa yang sama. Waktu tiba gelombang P dan S digunakan untuk memperkirakan jarak stasiun terhadap sumber gempa. Selanjutnya, informasi dari beberapa stasiun digunakan secara bersama-sama untuk memperoleh lokasi sumber yang paling sesuai.

### I — Input

Data yang diberikan berupa:

1. Latitude dan longitude stasiun.
2. Waktu tiba gelombang P.
3. Waktu tiba gelombang S.
4. Kecepatan gelombang P dan S atau model kecepatan yang digunakan.
5. Data elevasi stasiun apabila tersedia.

### K — Constraints

1. Jangan mengubah nilai asli data.
2. Jelaskan setiap persamaan yang digunakan.
3. Pastikan format waktu dan satuan jarak konsisten.
4. Jangan menentukan hiposenter hanya berdasarkan satu stasiun.
5. Jika terdapat residual besar, tampilkan dan jelaskan kemungkinan penyebabnya.
6. Jangan menganggap model kecepatan sebagai kondisi bumi yang sebenarnya jika model tersebut hanya merupakan pendekatan.

### O — Output

Hasil harus mencakup:

1. Tabel data stasiun dan waktu tiba.
2. Perhitungan selisih waktu S–P.
3. Perkiraan jarak stasiun terhadap sumber.
4. Estimasi latitude, longitude, dan kedalaman hiposenter.
5. Residual atau selisih antara waktu observasi dan waktu hasil model.
6. Visualisasi posisi stasiun dan hiposenter.
7. Interpretasi hasil dalam bentuk paragraf.
8. Kesimpulan dan keterbatasan.

### Q — Quality Criteria

Perhitungan harus menggunakan satuan yang konsisten, langkah penentuan lokasi dapat diikuti mahasiswa, hasil tidak dibuat-buat, dan ketidakpastian akibat picking maupun model kecepatan harus dijelaskan.

### V — Verification

Periksa kembali perhitungan S–P, jarak tiap stasiun, koordinat hasil, serta residual waktu. Pastikan solusi yang diperoleh memberikan kecocokan yang masuk akal terhadap data dari beberapa stasiun.

# PROMPT 2 — ANALISIS METODE MASW

## 1. Narasi Permasalahan

Metode Multichannel Analysis of Surface Waves (MASW) merupakan metode geofisika yang memanfaatkan karakteristik dispersi gelombang permukaan untuk memperoleh informasi mengenai kecepatan gelombang geser atau Vs bawah permukaan. Data MASW umumnya direkam menggunakan beberapa geofon dengan jarak tertentu sehingga menghasilkan pola gelombang yang dapat dianalisis dalam domain frekuensi dan bilangan gelombang.

Tantangan dalam analisis MASW adalah mengidentifikasi kurva dispersi secara tepat dan menghubungkannya dengan model Vs terhadap kedalaman. AI dapat digunakan sebagai alat bantu untuk mengenali pola dispersi, membantu proses picking kurva, serta mengevaluasi hasil inversi dengan tetap mempertimbangkan asumsi model.

## 2. Struktur Good Prompt

### R — Role

Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode gelombang permukaan dan analisis MASW**.

### T — Task

Analisis data MASW untuk mengidentifikasi kurva dispersi dan memperoleh model awal kecepatan gelombang geser terhadap kedalaman.

### C — Context

Data diperoleh dari susunan beberapa geofon yang merekam gelombang permukaan. Hasil transformasi data menghasilkan citra dispersi yang menunjukkan hubungan antara frekuensi dan kecepatan fase. Kurva dispersi tersebut kemudian digunakan sebagai dasar untuk memperkirakan struktur Vs bawah permukaan.

### I — Input

Data yang diberikan berupa:

1. Data rekaman tiap geofon.
2. Jarak antar-geofon.
3. Posisi sumber.
4. Sampling rate dan durasi rekaman.
5. Citra atau hasil spektrum dispersi apabila tersedia.

### K — Constraints

1. Jangan mengubah data mentah tanpa alasan.
2. Jelaskan proses dari data rekaman hingga kurva dispersi.
3. Bedakan antara kecepatan fase dan kecepatan grup.
4. Jangan memilih kurva dispersi hanya berdasarkan tampilan visual tanpa menjelaskan kriterianya.
5. Jika terdapat mode fundamental dan higher mode, jelaskan perbedaannya.
6. Jangan menginterpretasikan Vs sebagai jenis batuan secara pasti.

### O — Output

Hasil harus mencakup:

1. Deskripsi data MASW.
2. Penjelasan tahapan pengolahan.
3. Identifikasi kurva dispersi.
4. Tabel pasangan frekuensi dan kecepatan fase.
5. Model Vs terhadap kedalaman.
6. Interpretasi perubahan Vs secara vertikal.
7. Keterbatasan metode dan hasil.
8. Kesimpulan.

### Q — Quality Criteria

Hasil harus sesuai prinsip gelombang permukaan, satuan konsisten, proses picking dapat ditelusuri, dan interpretasi geologi tidak melebihi informasi yang diberikan oleh data Vs.

### V — Verification

Periksa apakah kurva dispersi yang dipilih konsisten, apakah model Vs mampu menjelaskan kurva observasi, serta apakah perubahan Vs terhadap kedalaman masuk akal berdasarkan kondisi bawah permukaan.

# PROMPT 3 — INTERPRETASI DATA GROUND PENETRATING RADAR (GPR)

## 1. Narasi Permasalahan

Ground Penetrating Radar (GPR) merupakan metode geofisika elektromagnetik yang menggunakan gelombang elektromagnetik berfrekuensi tinggi untuk memetakan kondisi bawah permukaan dangkal. Data GPR dapat memperlihatkan reflektor yang berkaitan dengan perubahan sifat elektromagnetik material, keberadaan lapisan, rongga, utilitas bawah tanah, maupun struktur dangkal lainnya.

Interpretasi radargram membutuhkan perhatian terhadap noise, refleksi berulang, perubahan amplitudo, dan bentuk reflektor. AI dapat membantu mengenali pola pada radargram, tetapi hasil interpretasi tetap perlu dikaitkan dengan prinsip propagasi gelombang elektromagnetik dan kondisi geologi setempat.

## 2. Struktur Good Prompt

### R — Role

Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode Ground Penetrating Radar dan interpretasi data bawah permukaan dangkal**.

### T — Task

Analisis radargram GPR untuk mengidentifikasi reflektor utama, memperkirakan kedalaman target, dan menjelaskan kemungkinan kondisi bawah permukaan berdasarkan pola refleksi.

### C — Context

Survei GPR dilakukan untuk memperoleh gambaran struktur bawah permukaan dangkal. Gelombang elektromagnetik yang dipancarkan antena akan mengalami refleksi ketika melewati batas material dengan sifat elektromagnetik yang berbeda.

### I — Input

Data yang diberikan berupa:

1. Radargram hasil survei.
2. Frekuensi antena.
3. Time window.
4. Kecepatan gelombang atau permitivitas relatif apabila tersedia.
5. Posisi lintasan dan informasi geologi pendukung.

### K — Constraints

1. Jangan menentukan kedalaman tanpa menjelaskan parameter kecepatan yang digunakan.
2. Jangan menganggap setiap reflektor sebagai batas lapisan geologi.
3. Bedakan reflektor geologi dan kemungkinan artefak/noise.
4. Jangan mengubah radargram asli tanpa menjelaskan proses pengolahan.
5. Jika permitivitas atau kecepatan tidak tersedia, jelaskan bahwa kedalaman bersifat perkiraan.

### O — Output

Hasil harus mencakup:

1. Penjelasan karakteristik radargram.
2. Identifikasi reflektor utama.
3. Estimasi kedalaman target.
4. Penjelasan hubungan amplitudo dan bentuk reflektor dengan kondisi bawah permukaan.
5. Identifikasi kemungkinan noise atau artefak.
6. Interpretasi geologi secara hati-hati.
7. Kesimpulan.

### Q — Quality Criteria

Interpretasi harus mengikuti prinsip propagasi gelombang elektromagnetik, penggunaan satuan harus benar, ketidakpastian kedalaman harus dijelaskan, dan kesimpulan harus dibedakan dari dugaan interpretasi.

### V — Verification

Periksa kembali konversi two-way travel time menjadi kedalaman, nilai kecepatan yang digunakan, posisi reflektor pada radargram, serta kesesuaiannya dengan informasi geologi pendukung.

# PROMPT 4 — ANALISIS MAGNETOTELLURIK (MT)

## 1. Narasi Permasalahan

Metode Magnetotellurik (MT) merupakan metode geofisika pasif yang memanfaatkan variasi alami medan elektromagnetik bumi untuk mengetahui distribusi resistivitas bawah permukaan. Metode ini dapat digunakan untuk mempelajari struktur bawah permukaan dari kedalaman dangkal hingga sangat dalam, bergantung pada rentang frekuensi dan kondisi pengukuran.

Dalam pengolahan MT, data umumnya dinyatakan dalam bentuk resistivitas semu dan fase terhadap frekuensi. Interpretasi membutuhkan pemahaman terhadap hubungan antara frekuensi, skin depth, dan sensitivitas kedalaman. AI dapat digunakan untuk membantu mengidentifikasi pola kurva, melakukan klasifikasi awal, dan membantu mengevaluasi model resistivitas.

## 2. Struktur Good Prompt

### R — Role

Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode elektromagnetik dan interpretasi data Magnetotellurik**.

### T — Task

Analisis data MT untuk menjelaskan variasi resistivitas terhadap frekuensi serta membuat interpretasi awal mengenai perubahan sifat listrik bawah permukaan.

### C — Context

Data MT diperoleh dari pengukuran variasi medan listrik dan magnet alami bumi. Hasil pengolahan dapat menghasilkan kurva resistivitas semu dan fase pada berbagai frekuensi. Frekuensi yang lebih rendah umumnya memberikan sensitivitas terhadap kedalaman yang lebih besar.

### I — Input

Data yang diberikan berupa:

1. Frekuensi.
2. Resistivitas semu.
3. Fase.
4. Komponen impedansi apabila tersedia.
5. Informasi geologi pendukung.

### K — Constraints

1. Gunakan prinsip dasar metode MT.
2. Jangan mengubah data asli.
3. Jelaskan hubungan frekuensi dengan kedalaman penetrasi.
4. Jangan menentukan jenis batuan secara pasti hanya dari resistivitas.
5. Bedakan data observasi dengan hasil model.
6. Jika terdapat data noisy, jelaskan ciri dan pengaruhnya terhadap interpretasi.

### O — Output

Hasil harus mencakup:

1. Tabel data frekuensi, resistivitas semu, dan fase.
2. Grafik log resistivitas semu terhadap frekuensi.
3. Grafik fase terhadap frekuensi.
4. Identifikasi zona resistif dan konduktif.
5. Penjelasan hubungan pola kurva dengan kemungkinan perubahan bawah permukaan.
6. Interpretasi awal model resistivitas.
7. Keterbatasan dan kesimpulan.

### Q — Quality Criteria

Analisis harus mempertahankan karakteristik data MT, menggunakan skala dan satuan yang tepat, membedakan observasi dengan interpretasi, serta menjelaskan ketidakpastian hasil.

### V — Verification

Periksa konsistensi antara kurva resistivitas dan fase, periksa data yang menyimpang, serta pastikan interpretasi kedalaman tidak dibuat hanya berdasarkan satu frekuensi.

# PROMPT 5 — ANALISIS DEFORMASI PERMUKAAN DENGAN InSAR

## 1. Narasi Permasalahan

Interferometric Synthetic Aperture Radar (InSAR) merupakan teknik penginderaan jauh yang dapat digunakan untuk mengukur perubahan posisi atau deformasi permukaan bumi melalui perbandingan fase sinyal radar dari beberapa akuisisi satelit. Teknik ini banyak dimanfaatkan untuk memantau deformasi akibat aktivitas tektonik, aktivitas vulkanik, penurunan tanah, maupun perubahan permukaan lainnya.

Data InSAR memiliki berbagai sumber ketidakpastian, seperti atmospheric delay, decorrelation, orbit, topografi, dan noise fase. Oleh karena itu, interpretasi deformasi perlu dilakukan secara hati-hati. AI dapat digunakan untuk membantu mengenali pola deformasi dan mengelompokkan area berdasarkan karakteristik perubahan, tetapi hasilnya tetap harus diverifikasi dengan data pendukung.

## 2. Struktur Good Prompt

### R — Role

Anda berperan sebagai seorang **geofisikawan yang ahli dalam penginderaan jauh geofisika dan analisis deformasi permukaan menggunakan InSAR**.

### T — Task

Analisis data deformasi InSAR untuk mengidentifikasi pola perubahan permukaan, menentukan area yang mengalami deformasi relatif besar, dan menjelaskan kemungkinan faktor geofisika yang berkaitan.

### C — Context

Data InSAR dapat menunjukkan perubahan fase yang kemudian dikonversi menjadi deformasi sepanjang line-of-sight (LOS). Hasil tersebut dapat digunakan untuk memetakan pola deformasi permukaan pada suatu wilayah dalam periode tertentu.

### I — Input

Data yang diberikan berupa:

1. Peta atau raster deformasi InSAR.
2. Koordinat lokasi.
3. Periode akuisisi data.
4. Nilai deformasi LOS.
5. Data elevasi atau topografi apabila tersedia.
6. Data GPS atau pengamatan lapangan apabila tersedia.

### K — Constraints

1. Jangan mengubah nilai deformasi asli.
2. Jelaskan bahwa deformasi LOS bukan selalu perpindahan vertikal murni.
3. Bedakan pola deformasi dengan artefak atmosfer atau noise.
4. Jangan menentukan penyebab deformasi secara pasti tanpa data pendukung.
5. Gunakan periode waktu yang sesuai dengan dataset.
6. Jika data validasi tidak tersedia, nyatakan keterbatasannya.

### O — Output

Hasil harus mencakup:

1. Deskripsi dataset InSAR.
2. Peta atau visualisasi deformasi.
3. Identifikasi zona deformasi positif dan negatif.
4. Nilai maksimum dan minimum deformasi beserta lokasinya.
5. Analisis pola spasial deformasi.
6. Perbandingan dengan topografi atau data geologi apabila tersedia.
7. Interpretasi kemungkinan proses geofisika.
8. Keterbatasan dan kesimpulan.

### Q — Quality Criteria

Hasil harus membedakan pengamatan dengan interpretasi, mempertahankan satuan deformasi, memperhatikan arah LOS, serta tidak mengklaim penyebab deformasi secara pasti tanpa bukti pendukung.

### V — Verification

Periksa kembali satuan dan rentang deformasi, lokasi nilai ekstrem, konsistensi pola spasial, serta bandingkan dengan data validasi apabila tersedia. Identifikasi pula apakah pola tertentu berpotensi berasal dari atmosfer atau noise.
